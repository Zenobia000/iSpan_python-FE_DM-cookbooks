我們將對每個類別特徵進行分析，並選擇最適合它的編碼方法。

```python
# 導入必要的函式庫
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

## 1. 載入並初步清理資料

我們先載入資料，並進行最基礎的清理，例如填充一些已知的缺失值。

```python
# 載入資料
path = '../../../datasets/raw/titanic/Titanic-Dataset.csv'
try:
    df = pd.read_csv(path)
    print("成功載入 Titanic 資料集!")
except FileNotFoundError:
    print(f"在 '{path}' 找不到 train.csv。")
    df = pd.DataFrame()

# 建立一個工作副本
df_processed = df.copy()

# 簡單填充 'Embarked' 和 'Age' 的缺失值（這裡使用簡單策略，更複雜的會在模組3討論）
df_processed = df_processed.assign(
    Embarked=df_processed['Embarked'].fillna(df_processed['Embarked'].mode()[0]),
    Age=df_processed['Age'].fillna(df_processed['Age'].median())
)

# 刪除 'Cabin' (缺失過多) 和其他暫不使用的欄位
df_processed.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1, inplace=True)

print("\n初步清理後的資料狀態:")
df_processed.info()

```

## 2. 逐一分析與編碼類別特徵

我們的目標是將 `Sex`, `Embarked` 轉換為數值格式。

### 2.1 `Sex` 特徵

- **類型**: 名目型 (Nominal)
- **基數**: 低 (只有 'male', 'female' 兩個值)
- **策略**: 這是獨熱編碼的完美應用場景。但由於它只有兩個值，使用標籤編碼 (0/1) 也能達到相同的效果，且不會引入錯誤的順序性，同時更節省空間。我們這裡選擇更簡潔的標籤編碼。

```python
# 方法一：使用 LabelEncoder
le_sex = LabelEncoder()
df_processed['Sex_LabelEncoded'] = le_sex.fit_transform(df_processed['Sex'])
print("Sex 使用 LabelEncoder 後:")
print(df_processed[['Sex', 'Sex_LabelEncoded']].head())

# 方法二：使用 map (更直觀)
sex_mapping = {'male': 0, 'female': 1}
df_processed['Sex'] = df_processed['Sex'].map(sex_mapping)

# 刪除臨時欄位
df_processed.drop('Sex_LabelEncoded', axis=1, inplace=True, errors='ignore')

print("\nSex 使用 map 轉換後:")
print(df_processed.head())

```

### 2.2 `Embarked` 特徵

- **類型**: 名目型 (Nominal)
- **基數**: 低 (S, C, Q 三個值)
- **策略**: 由於是無序的名目特徵，且我們可能要將其用於線性模型，最安全、最標準的選擇是 **獨熱編碼**。

```python
# 使用 pd.get_dummies 進行獨熱編碼
embarked_onehot = pd.get_dummies(df_processed['Embarked'], prefix='Embarked')

# 將編碼結果與原 DataFrame 合併
df_processed = pd.concat([df_processed, embarked_onehot], axis=1)

# 刪除原始的 'Embarked' 欄位
df_processed.drop('Embarked', axis=1, inplace=True)

print("Embarked 獨熱編碼後:")
display(df_processed.head())


```

## 3. 創建一個新的順序特徵

雖然我們沒有現成的順序特徵，但我們可以從 `Age` 創建一個。例如，我們可以將年齡分箱 (Binning) 成幾個年齡段。

- **類型**: 順序型 (Ordinal)
- **策略**: 手動定義映射關係，然後使用標籤編碼。

```python
# 將 Age 分箱
bins = [0, 12, 18, 60, np.inf] # 定義邊界
labels = ['Child', 'Teenager', 'Adult', 'Senior'] # 定義標籤
df_processed['AgeGroup'] = pd.cut(df_processed['Age'], bins=bins, labels=labels, right=False)

# 定義順序映射
age_group_mapping = {'Child': 0, 'Teenager': 1, 'Adult': 2, 'Senior': 3}

# 進行編碼
df_processed['AgeGroup_Encoded'] = df_processed['AgeGroup'].map(age_group_mapping)

print("創建並編碼 AgeGroup 後:")
display(df_processed[['Age', 'AgeGroup', 'AgeGroup_Encoded']].head())


```

## 4. 最終檢視

讓我們看看最終處理完成的資料集。

```python
print("所有類別變數編碼完成後的資料集:")
df_processed.info()

display(df_processed.head())

```

**結果**:
我們成功地將原始資料集中的所有特徵都轉換為了數值型態，包括：
- `Survived`, `Pclass`, `Age`, `SibSp`, `Parch`, `Fare`: 原始數值或填充後的數值。
- `Sex`: 經過標籤編碼 (0/1)。
- `Embarked_C`, `Embarked_Q`, `Embarked_S`: 經過獨熱編碼。
- `AgeGroup_Encoded`: 手動創建並進行順序編碼。

這個 `df_processed` DataFrame 現在已經是一個完全數值化的、乾淨的資料集，可以直接輸入到任何機器學習模型中進行訓練了。

## 5. 特徵工程有效嗎？Gradient Boosting 比較

`Survived` 是 0/1，這是 **二元分類**。我們用同一個 `GradientBoostingClassifier`、同一組訓練／測試切分，只改特徵：

| 策略 | 用哪些欄 | 做了什麼 |
|------|----------|----------|
| 無特徵工程 | `Pclass`, `Age`, `SibSp`, `Parch`, `Fare` | 不用 `Sex`、`Embarked`（字串，sklearn GBT 無法直接吃）。`Age` 只做訓練能跑所需的中位數填補 |
| 本筆記本編碼 | 上一節的 `df_processed` | `Sex` 標籤編碼、`Embarked` 獨熱、`Age` 分箱後順序編碼 |

差異只來自特徵，不來自模型或切分。評估指標以 **Accuracy** 為主；存活率不平衡（約 38% 存活），表內一併列出 Precision / Recall / F1 / ROC-AUC。

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)

y = df['Survived']

# --- 無特徵工程：只用原始數值欄 ---
X_raw = df[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].copy()
X_raw['Age'] = X_raw['Age'].fillna(df['Age'].median())

# --- 本筆記本編碼：沿用前面做好的 df_processed ---
X_fe = df_processed.drop(columns=['Survived', 'AgeGroup']).copy()
X_fe['AgeGroup_Encoded'] = X_fe['AgeGroup_Encoded'].astype(int)

print('無特徵工程欄位:', X_raw.columns.tolist())
print('本筆記本編碼欄位:', X_fe.columns.tolist())

# 同一批乘客、同一隨機種子、分層切分（存活比例與全資料一致）
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.2, random_state=42, stratify=y
)
X_train_fe, X_test_fe, _, _ = train_test_split(
    X_fe, y, test_size=0.2, random_state=42, stratify=y
)


def evaluate_gbt(strategy_name, X_train, X_test):
    model = GradientBoostingClassifier(random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]
    return {
        '特徵策略': strategy_name,
        '特徵數': X_train.shape[1],
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_proba),
    }, model


row_raw, _ = evaluate_gbt('無特徵工程（僅原始數值欄）', X_train_raw, X_test_raw)
row_fe, model_fe = evaluate_gbt('本筆記本編碼流程', X_train_fe, X_test_fe)

results_df = pd.DataFrame([row_raw, row_fe])
metric_cols = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC-AUC']
display(results_df.style.format({col: '{:.4f}' for col in metric_cols}))

print('\n本筆記本編碼流程的特徵重要性（由高到低）：')
importance = (
    pd.Series(model_fe.feature_importances_, index=X_fe.columns)
    .sort_values(ascending=False)
    .to_frame('importance')
)
display(importance.style.format({'importance': '{:.4f}'}))
```

**怎麼讀這張表**

- 兩列是 **同一個分類器、同一組測試乘客**。Accuracy 從約 **0.63 升到約 0.80**，差的是特徵，不是模型。
- 跳升的主因是把 `Sex` 編碼進來。鐵達尼號的存活與性別高度相關；無特徵工程時模型根本看不到這欄。
- `AgeGroup_Encoded` 的重要性通常很低：原始 `Age` 已經在特徵裡，分箱只是同一資訊的粗版本。
- Accuracy 不是唯一指標。存活是少數類，看 Recall / F1 / ROC-AUC 比較能判斷「真的救到多少存活者」。

## 總結

在這個案例中，我們針對一個真實資料集，根據每個類別特徵的具體情況制定了不同的編碼策略：

- 對於二元類別特徵 (`Sex`)，標籤編碼是一種簡潔高效的選擇。
- 對於低基數的名目特徵 (`Embarked`)，獨熱編碼是最標準、最不會引入錯誤資訊的方法。
- 我們還展示了如何通過分箱來 **創建** 自己的順序特徵，並對其進行有序的標籤編碼。

最後用同一個 Gradient Boosting 分類器做對照：不編碼時模型看不到 `Sex` / `Embarked`，測試 Accuracy 明顯較低；套用本筆記本的編碼後，存活預測準確度上升。這說明編碼策略要依特徵性質來選，而且選完要用模型驗證它有沒有真的幫忙。

這個流程展示了在實際工作中，我們通常不會只用一種編碼方法，而是會根據特徵和模型的需要，靈活地組合使用多種技術。
